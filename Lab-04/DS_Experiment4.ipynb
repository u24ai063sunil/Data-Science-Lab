## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
%matplotlib inline
```

## 2. Load the Titanic Dataset

The dataset is loaded using Seaborn's built-in dataset loader. If Seaborn's online dataset
cannot be reached (no internet access), the dataset is loaded from a public backup CSV URL instead.

```python
try:
    df = sns.load_dataset('titanic')
    print("Loaded Titanic dataset from Seaborn.")
except Exception as e:
    print("Seaborn load failed, falling back to CSV URL:", e)
    url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
    df = pd.read_csv(url)
    print("Loaded Titanic dataset from CSV URL.")

print("Shape of dataset:", df.shape)
df.head()
```

## 3. Select Required Variables

For this experiment we only need the **age** and **survived** columns.

```python
data = df[['age', 'survived']].copy()
data.head()
```

## 4. Remove Missing Values

```python
print("Missing values before cleaning:\n", data.isnull().sum())

data_clean = data.dropna(subset=['age'])

print("\nMissing values after cleaning:\n", data_clean.isnull().sum())
print("\nShape after cleaning:", data_clean.shape)
```

## 5. Separate Passengers into Survived and Not-Survived Groups

```python
survived_age = data_clean[data_clean['survived'] == 1]['age']
not_survived_age = data_clean[data_clean['survived'] == 0]['age']

print("Number of survivors (with age):", len(survived_age))
print("Number of non-survivors (with age):", len(not_survived_age))

print("\nMean age (Survived):", round(survived_age.mean(), 2))
print("Mean age (Not Survived):", round(not_survived_age.mean(), 2))
```

## 6. Formulate the Hypotheses

- **Null Hypothesis (H0):** There is no significant difference in the average age between passengers who survived and those who did not.
- **Alternative Hypothesis (H1):** There is a significant difference in the average age between passengers who survived and those who did not.

Significance level: **alpha = 0.05**

## 7. Perform the Independent Sample t-test

```python
t_stat, p_value = stats.ttest_ind(survived_age, not_survived_age, equal_var=False)

print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_value, 4))
```

## 8. Interpret the Result

```python
alpha = 0.05

if p_value < alpha:
    print(f"p-value ({p_value:.4f}) < alpha ({alpha}) --> Reject the Null Hypothesis (H0).")
    print("Conclusion: There IS a statistically significant difference in average age between survivors and non-survivors.")
else:
    print(f"p-value ({p_value:.4f}) >= alpha ({alpha}) --> Fail to Reject the Null Hypothesis (H0).")
    print("Conclusion: There is NO statistically significant difference in average age between survivors and non-survivors.")
```

## 9. Visualization

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='survived', y='age', data=data_clean, palette=['#e74c3c', '#2ecc71'])
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.xlabel('Survival Status')
plt.ylabel('Age')
plt.title('Age Distribution by Survival Status')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(survived_age, color='#2ecc71', label='Survived', kde=True, alpha=0.5)
sns.histplot(not_survived_age, color='#e74c3c', label='Not Survived', kde=True, alpha=0.5)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution: Survived vs Not Survived')
plt.legend()
plt.show()
```

## Result

Statistical hypothesis testing was successfully performed using Python. The hypothesis was tested
using an Independent Sample t-test, and the conclusion was drawn based on the calculated p-value.

## Conclusion

The experiment demonstrated the application of statistical hypothesis testing using Python. The
Independent Sample t-test was used to compare the mean age of survived and non-survived
passengers. Based on the p-value, an appropriate statistical conclusion was made regarding the null
hypothesis. Hypothesis testing is an important tool for making data-driven decisions and validating
assumptions in data science.

## Exercise 1: Perform a t-test using the Fare variable instead of Age

**H0:** There is no significant difference in the average fare between passengers who survived and those who did not.
**H1:** There is a significant difference in the average fare between passengers who survived and those who did not.

```python
fare_data = df[['fare', 'survived']].dropna()

survived_fare = fare_data[fare_data['survived'] == 1]['fare']
not_survived_fare = fare_data[fare_data['survived'] == 0]['fare']

t_stat_fare, p_value_fare = stats.ttest_ind(survived_fare, not_survived_fare, equal_var=False)

print("Mean Fare (Survived):", round(survived_fare.mean(), 2))
print("Mean Fare (Not Survived):", round(not_survived_fare.mean(), 2))
print("t-statistic:", round(t_stat_fare, 4))
print("p-value:", round(p_value_fare, 4))

if p_value_fare < 0.05:
    print("\nResult: Reject H0 -> Significant difference in average fare between the two groups.")
else:
    print("\nResult: Fail to reject H0 -> No significant difference in average fare between the two groups.")
```

## Exercise 4: Apply another statistical test such as the Chi-Square test

Here we test whether **Passenger Class (pclass)** and **Survival (survived)** are independent of each other.

**H0:** Passenger class and survival status are independent (no association).
**H1:** Passenger class and survival status are associated (not independent).

```python
contingency_table = pd.crosstab(df['pclass'], df['survived'])
print("Contingency Table (Pclass vs Survived):")
print(contingency_table)

chi2_stat, p_value_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print("\nChi-Square Statistic:", round(chi2_stat, 4))
print("Degrees of Freedom:", dof)
print("p-value:", round(p_value_chi2, 6))

if p_value_chi2 < 0.05:
    print("\nResult: Reject H0 -> Passenger class and survival are significantly associated.")
else:
    print("\nResult: Fail to reject H0 -> No significant association between passenger class and survival.")
```

## Exercise 5: State the null and alternative hypotheses for a different variable

**Variable chosen: Sex (Gender) vs Survival**

- **Null Hypothesis (H0):** There is no significant association between a passenger's gender and their survival status (gender and survival are independent).
- **Alternative Hypothesis (H1):** There is a significant association between a passenger's gender and their survival status (gender and survival are not independent).

This can be tested using a Chi-Square test of independence, as shown below (for verification).

```python
contingency_sex = pd.crosstab(df['sex'], df['survived'])
print("Contingency Table (Sex vs Survived):")
print(contingency_sex)

chi2_sex, p_value_sex, dof_sex, expected_sex = stats.chi2_contingency(contingency_sex)

print("\nChi-Square Statistic:", round(chi2_sex, 4))
print("p-value:", round(p_value_sex, 6))

if p_value_sex < 0.05:
    print("\nResult: Reject H0 -> Gender and survival status are significantly associated.")
else:
    print("\nResult: Fail to reject H0 -> No significant association between gender and survival status.")
```
