## Theory / Background

**Exploratory Data Analysis (EDA)** is the process of examining a dataset to understand its structure, detect patterns, identify missing values, and discover relationships between variables before applying machine learning or statistical models.

**Descriptive statistics** summarize the main characteristics of data using numerical measures such as mean, median, mode, standard deviation, minimum, maximum, and quartiles. These measures help understand the distribution and variability of the dataset.

**Data visualization** represents data graphically, making it easier to identify trends, patterns, and outliers. Python libraries such as Matplotlib and Seaborn provide various visualization techniques including histograms, bar charts, box plots, scatter plots, and heatmaps.

EDA helps improve data quality and supports better decision-making by providing meaningful insights into the dataset.

## Step 1: Import Required Libraries

We import Pandas for data handling, NumPy for numerical operations, and Matplotlib/Seaborn for visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
sns.set(style="whitegrid")
%matplotlib inline

print("Libraries imported successfully.")
```

## Step 2: Load the Titanic Dataset

The Titanic dataset is directly available through Seaborn's built-in dataset repository, so we load it using `sns.load_dataset('titanic')`.

```python
# Load the Titanic dataset from Seaborn's built-in datasets
titanic = sns.load_dataset('titanic')

# Display the first 5 records
titanic.head()
```

## Step 3: Display Dataset Information

We check the shape, column names, data types, and general structure of the dataset.

```python
# Shape of the dataset (rows, columns)
print("Shape of dataset:", titanic.shape)

# Column names
print("\nColumn names:\n", titanic.columns.tolist())
```

```python
# Detailed information about the dataset (data types, non-null counts)
titanic.info()
```

## Step 4: Check for Missing Values

Identifying missing values is a crucial step in EDA since missing data can affect statistical analysis and model performance.

```python
# Count of missing values in each column
missing_values = titanic.isnull().sum()
print("Missing values per column:\n")
print(missing_values)
```

```python
# Percentage of missing values per column
missing_percent = (titanic.isnull().sum() / len(titanic)) * 100
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)
print("Percentage of missing values:\n")
print(missing_percent)
```

```python
# Visualizing missing values using a heatmap
plt.figure(figsize=(10, 5))
sns.heatmap(titanic.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

## Step 5: Descriptive Statistics

We use the `describe()` function to calculate summary statistics (mean, median, standard deviation, min, max, quartiles) for numerical variables.

```python
# Descriptive statistics for numerical columns
titanic.describe()
```

```python
# Descriptive statistics including categorical columns
titanic.describe(include='all')
```

```python
# Individual statistical measures for the 'age' column
print("Mean Age:", titanic['age'].mean())
print("Median Age:", titanic['age'].median())
print("Mode Age:", titanic['age'].mode()[0])
print("Standard Deviation of Age:", titanic['age'].std())
print("Minimum Age:", titanic['age'].min())
print("Maximum Age:", titanic['age'].max())
```

## Step 6: Visualizing Categorical Variables

Bar charts are used to visualize the frequency distribution of categorical variables such as `sex`, `pclass`, `embarked`, and `survived`.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.countplot(x='survived', data=titanic, ax=axes[0, 0])
axes[0, 0].set_title('Survival Count (0 = No, 1 = Yes)')

sns.countplot(x='sex', data=titanic, ax=axes[0, 1])
axes[0, 1].set_title('Gender Distribution')

sns.countplot(x='pclass', data=titanic, ax=axes[1, 0])
axes[1, 0].set_title('Passenger Class Distribution')

sns.countplot(x='embarked', data=titanic, ax=axes[1, 1])
axes[1, 1].set_title('Port of Embarkation Distribution')

plt.tight_layout()
plt.show()
```

```python
# Survival count grouped by gender
plt.figure(figsize=(6, 4))
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title('Survival Count by Gender')
plt.show()
```

```python
# Survival count grouped by passenger class
plt.figure(figsize=(6, 4))
sns.countplot(x='pclass', hue='survived', data=titanic)
plt.title('Survival Count by Passenger Class')
plt.show()
```

## Step 7: Visualizing Numerical Variables

Histograms show the distribution/frequency of numerical variables, while box plots help identify outliers.

```python
# Histogram of Age
plt.figure(figsize=(7, 5))
sns.histplot(titanic['age'].dropna(), bins=30, kde=True, color='steelblue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Histogram of Fare
plt.figure(figsize=(7, 5))
sns.histplot(titanic['fare'].dropna(), bins=30, kde=True, color='seagreen')
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
# Box plot for Age to detect outliers
plt.figure(figsize=(6, 4))
sns.boxplot(x=titanic['age'], color='orange')
plt.title('Box Plot of Age (Outlier Detection)')
plt.show()
```

```python
# Box plot for Fare to detect outliers
plt.figure(figsize=(6, 4))
sns.boxplot(x=titanic['fare'], color='tomato')
plt.title('Box Plot of Fare (Outlier Detection)')
plt.show()
```

```python
# Box plot of Fare grouped by Passenger Class
plt.figure(figsize=(7, 5))
sns.boxplot(x='pclass', y='fare', data=titanic)
plt.title('Fare Distribution across Passenger Classes')
plt.show()
```

## Step 8: Scatter Plot

Scatter plots help visualize relationships between two numerical variables, such as Age and Fare.

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(x='age', y='fare', hue='survived', data=titanic, alpha=0.7)
plt.title('Age vs Fare (colored by Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

## Step 9: Correlation Heatmap

A correlation heatmap displays the strength of the linear relationship between numerical variables. Values close to +1 indicate strong positive correlation, values close to -1 indicate strong negative correlation, and values near 0 indicate little to no correlation.

```python
# Select only numerical columns for correlation
numeric_cols = titanic.select_dtypes(include=[np.number])

plt.figure(figsize=(9, 7))
corr_matrix = numeric_cols.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

- **Passenger class** also influenced survival — 1st class passengers had a noticeably higher survival rate than 3rd class passengers.

- **Box plots** for Age and Fare reveal several outliers, especially in the Fare column, corresponding to passengers who paid unusually high ticket prices.
- The **correlation heatmap** shows a moderate positive correlation between `fare` and `survived`, and a negative correlation between `pclass` and `fare` (since class 1 = highest fare).

```python
# Exercise 2: Histogram of the Age variable
plt.figure(figsize=(7, 5))
sns.histplot(titanic['age'].dropna(), bins=25, kde=True, color='purple')
plt.title('Exercise 2: Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Exercise 3: Box plot for Fare to identify outliers
plt.figure(figsize=(6, 4))
sns.boxplot(x=titanic['fare'], color='gold')
plt.title('Exercise 3: Box Plot of Fare')
plt.show()

# Identify outliers using the IQR method
Q1 = titanic['fare'].quantile(0.25)
Q3 = titanic['fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = titanic[(titanic['fare'] < lower_bound) | (titanic['fare'] > upper_bound)]
print(f"Number of outliers in Fare: {len(outliers)}")
```

```python
# Exercise 4: Heatmap showing correlation between numerical variables
plt.figure(figsize=(9, 7))
sns.heatmap(titanic.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Exercise 4: Correlation Heatmap')
plt.show()
```

Looking at the Fare box plot (Exercise 3), most passengers paid a relatively low fare (concentrated between roughly \$7 and \$31), but there are several outliers with very high fares (above \$100, with a few exceeding \$500). This right-skew indicates that a small number of wealthy first-class passengers paid disproportionately high fares compared to the rest of the passengers, which is consistent with the class-based pricing structure aboard the Titanic.
